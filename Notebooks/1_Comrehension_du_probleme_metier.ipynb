<b> <a>  INTRODUCTION

Le marché immobilier repose sur l'estimation  de la valeur des biens afin de faciliter les décisions d'achat, de vente et d'investissement. Cependant, determiner le prix approprié d'une maison peut etre complexe, car sa valeur dépend de plusieurs caractéridtiques telles que sa supérficie, son emplacement, le nombre de chambres, le nombre de salles de bain et d'autres facteurs.

Dans ce contexte, ce projet consiste à exploiter des techniques de Machine Learning pour développer un modèle capable d'estimer le prix d'une maison à partir  de ses caracteristiques.

<b> <a>  PROBLEME METIER  

Les acteurs du marché immobilier ont des difficultés à determiner rapidement et précisement le prix approprié d'un bien immobilier.

Une mauvaise estimation peut entrainer une vente trop lente lorsque le prix est trop élevè, ou une perte financière
lorsque le prix est sous-évalué.

<b> <a>  OBJECTIF DU PROJET

Developper un modèle de Machine Learning capable de prédire le prix d'une maison à partir de ses caracteristiques,
afin de fournir une estimation fiable pouvant aider à la prise de décision
